```python
def performance(score):
    if score >= 85:
        return "Excellent"
    elif score >= 70:
        return "Good"
    elif score >= 50:
        return "Average"
    else:
        return "Poor"

df["Performance"] = df["Exam_Score"].apply(performance)

print(df)
```

```python
df["Total_Score"] = df["Exam_Score"] + df["Quiz_Score"]
print(df)
```

```python
def Study_Level(Study_hours):
  if Study_hours >= 4:
    return "High"
  elif Study_hours >=2:
    return "Medium"
  else:
    return "Low"


df["Study_Level"] = df["Study_Hours"].apply(Study_Level)
print(df)
```

```python
def Attendance_status(Attendance):
  if Attendance >=85:
    return "Good"
  elif Attendance >=70:
    return "Average"
  else:
    return "Poor"

df["Attendance_Status"] = df["Attendance"].apply(Attendance_status)
print(df)
```
