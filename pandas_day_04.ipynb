```python
import pandas as pd
df = pd.read_csv("/content/employee_dataset.csv")
departments = pd.read_csv("/content/departments.csv")
employees = pd.read_csv("/content/new_employees.csv")
print(df)
```

```python
df = df.drop(columns=["City"])
```

```python
df.isnull().sum()
```

```python
df.dropna()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df["Salary"] = df["Salary"].fillna(0)
```

```python
df["Joining_Date"] = df["Joining_Date"].fillna("Unknown")
```

```python
df["Gender"] = df["Gender"].replace("M","Male")
df["Gender"] = df["Gender"].replace("male","Male")
df["Gender"] = df["Gender"].replace("F","Female")
```

```python
df = df.rename(columns={"Employee_ID": "ID", "Joining_Date": "Join_Date", "Department": "Dept"})
```
