```python
def performance(score):
  if score >= 85:
    return "Excellent"
  elif score >= 70:
    return "Good"
  elif score >= 50:
    return "Average"
  else:
    return "Poor"
df["Performance"] = df["Exam_Score"].apply(performance)
print(df)
```

```python
df["Total_Score"] = df["Assignment_Score"] + df["Quiz_Score"] + df["Exam_Score"]
```

```python
df.sort_values("Total_Score",ascending=False)["Name"].head(5)
```

```python
corr = df[["Study_Hours" , "Attendance", "Exam_Score"]].corr()
corr = corr.where(corr < 1)
pair = corr.stack().idxmax()
highest = corr.stack().max()
print("Pair", pair)
print("Correlation", highest)
```
